```python
import pandas as pd

# Definimos las 5 tablas que limpiamos en la Fase 1
archivos_procesados = [
    'Match_LaLiga.csv',
    'Team_LaLiga.csv',
    'Team_Attributes_LaLiga.csv',
    'Player_LaLiga.csv',
    'Player_Attributes_LaLiga.csv'
]

datos_diccionario_global = []

# Generamos la iteracion sobre cada archivo
for archivo in archivos_procesados:
    ruta = f'../data/processed/{archivo}'
    
    try:
        # Leemos el archivo
        df = pd.read_csv(ruta)
        nombre_tabla = archivo.replace('_LaLiga.csv', '') # Limpiamos el nombre para que se lea mejor
        
        # Iteramos sobre las columnas de ese archivo
        for columna in df.columns:
            tipo = str(df[columna].dtype)
            nulos = df[columna].isnull().sum()
            pct_nulos = (nulos / len(df)) * 100
            
            # Extraemos un ejemplo real
            ejemplo = df[columna].dropna().iloc[0] if not df[columna].dropna().empty else "Sin datos"
            
            datos_diccionario_global.append({
                'Tabla': nombre_tabla,
                'Campo': columna,
                'Tipo_Python': tipo,
                'Nulos': nulos,
                'Porcentaje_Nulos(%)': round(pct_nulos, 2),
                'Ejemplo_Valor': ejemplo
            })
            
    except FileNotFoundError:
        print(f"Advertencia: No se encontró el archivo {ruta}. Verifica que esté en la carpeta.")

# Convertimos todo a un DataFrame para nuestro diccionario
df_diccionario_global = pd.DataFrame(datos_diccionario_global)

# Exportamos el diccionario unificado
df_diccionario_global.to_csv('../data/processed/Diccionario_Global.csv', index=False)

print(f"Total de variables mapeadas en todo el proyecto: {len(df_diccionario_global)}")
display(df_diccionario_global)
```
