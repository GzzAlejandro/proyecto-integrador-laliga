```python
# 1. PREPARACIÓN DE DATOS
# Asumiendo que df_match_clean y df_team_clean ya están cargados en tu entorno
# Creamos la columna de goles totales por partido
df_match_clean['total_goals'] = df_match_clean['home_team_goal'] + df_match_clean['away_team_goal']
```

```python
# 2. ANÁLISIS GLOBAL DE GOLES
# Calculamos Media, Mediana, Moda, Varianza y Desviación Estándar
metrics = ['home_team_goal', 'away_team_goal', 'total_goals']
global_stats = []

for col in metrics:
    stats = {
        'Metrica': col,
        'Suma_Total': df_match_clean[col].sum(),
        'Media': round(df_match_clean[col].mean(), 3),
        'Mediana': df_match_clean[col].median(),
        'Moda': df_match_clean[col].mode()[0], # mode() devuelve una serie, tomamos el primer valor
        'Varianza': round(df_match_clean[col].var(), 3),
        'Desviacion_Estandar': round(df_match_clean[col].std(), 3),
        'Minimo': df_match_clean[col].min(),
        'Maximo': df_match_clean[col].max()
    }
    global_stats.append(stats)

df_global_stats = pd.DataFrame(global_stats)
```

```python
# 3. ANÁLISIS POR TEMPORADA (Evolución temporal)
# Agrupamos por 'season' para ver cómo cambian los goles a lo largo de los años
df_season_stats = df_match_clean.groupby('season').agg(
    Partidos_Jugados=('id', 'count'),
    Goles_Locales_Suma=('home_team_goal', 'sum'),
    Goles_Visitantes_Suma=('away_team_goal', 'sum'),
    Goles_Totales_Suma=('total_goals', 'sum'),
    Promedio_Goles_Partido=('total_goals', 'mean'),
    Desv_Est_Goles=('total_goals', 'std')
).reset_index()

df_season_stats['Promedio_Goles_Partido'] = df_season_stats['Promedio_Goles_Partido'].round(2)
df_season_stats['Desv_Est_Goles'] = df_season_stats['Desv_Est_Goles'].round(2)
```

```python
# 4. ANÁLISIS POR EQUIPO
# Para hacer esto, necesitamos cruzar los IDs de los equipos con sus nombres usando df_team_clean

# 4.1 Rendimiento como Local (Home)
df_home = pd.merge(df_match_clean, df_team_clean[['team_api_id', 'team_long_name']], 
                   left_on='home_team_api_id', right_on='team_api_id', how='left')

df_team_home_stats = df_home.groupby('team_long_name').agg(
    Partidos_Local=('id', 'count'),
    Goles_A_Favor_Suma=('home_team_goal', 'sum'),
    Goles_A_Favor_Media=('home_team_goal', 'mean'),
    Goles_A_Favor_Mediana=('home_team_goal', 'median'),
    Goles_A_Favor_Desv=('home_team_goal', 'std'),
    Goles_En_Contra_Suma=('away_team_goal', 'sum'), # Como local, los goles del visitante son en contra
    Goles_En_Contra_Media=('away_team_goal', 'mean')
).reset_index().rename(columns={'team_long_name': 'Equipo'})

# 4.2 Rendimiento como Visitante (Away)
df_away = pd.merge(df_match_clean, df_team_clean[['team_api_id', 'team_long_name']], 
                   left_on='away_team_api_id', right_on='team_api_id', how='left')

df_team_away_stats = df_away.groupby('team_long_name').agg(
    Partidos_Visitante=('id', 'count'),
    Goles_A_Favor_Suma=('away_team_goal', 'sum'), # Como visitante, sus goles son 'away'
    Goles_A_Favor_Media=('away_team_goal', 'mean'),
    Goles_A_Favor_Mediana=('away_team_goal', 'median'),
    Goles_A_Favor_Desv=('away_team_goal', 'std'),
    Goles_En_Contra_Suma=('home_team_goal', 'sum'), 
    Goles_En_Contra_Media=('home_team_goal', 'mean')
).reset_index().rename(columns={'team_long_name': 'Equipo'})

# 4.3 Consolidado Total por Equipo (Opcional pero muy útil para Power BI)
# Sumamos las métricas de local y visitante
df_team_total = pd.merge(df_team_home_stats, df_team_away_stats, on='Equipo')
df_team_total['Goles_A_Favor_Totales'] = df_team_total['Goles_A_Favor_Suma_x'] + df_team_total['Goles_A_Favor_Suma_y']
df_team_total['Goles_En_Contra_Totales'] = df_team_total['Goles_En_Contra_Suma_x'] + df_team_total['Goles_En_Contra_Suma_y']
df_team_total['Diferencia_Goles'] = df_team_total['Goles_A_Favor_Totales'] - df_team_total['Goles_En_Contra_Totales']

# Limpiamos el dataframe final de equipos para que sea más legible
df_team_total = df_team_total[['Equipo', 'Goles_A_Favor_Totales', 'Goles_En_Contra_Totales', 'Diferencia_Goles']]
df_team_total = df_team_total.sort_values(by='Goles_A_Favor_Totales', ascending=False)
```

```python
# 5. EXPORTACIÓN A EXCEL PARA POWER BI
# Exportamos cada DataFrame a una hoja diferente dentro del mismo archivo Excel
ruta_exportacion = r'C:\Users\edgar\Downloads\proyecto-integrador-laliga\data\processed\clean\Analisis_Goles_LaLiga_metricas_macro_global.xlsx'

with pd.ExcelWriter(ruta_exportacion, engine='openpyxl') as writer:
    df_global_stats.to_excel(writer, sheet_name='Stats_Globales', index=False)
    df_season_stats.to_excel(writer, sheet_name='Stats_Temporada', index=False)
    df_team_home_stats.to_excel(writer, sheet_name='Stats_Local', index=False)
    df_team_away_stats.to_excel(writer, sheet_name='Stats_Visitante', index=False)
    df_team_total.to_excel(writer, sheet_name='Stats_Total_Equipo', index=False)
    # También puedes exportar el dataframe original con la nueva columna 'total_goals'
    df_match_clean.to_excel(writer, sheet_name='Match_Data_Completo', index=False)

print(f"Análisis completado y exportado con éxito a: {ruta_exportacion}")
```

### El script generó el desglose en las demás pestañas del archivo de Excel para darte los diferentes niveles de profundidad:

Stats_Globales: El comportamiento macro histórico de toda La Liga (todos los partidos combinados).


Stats_Temporada: El desglose año por año, para ver si en una temporada específica subió o bajó el promedio de goles.

Stats_Total_Equipo (junto con Stats_Local y Stats_Visitante): El desglose equipo por equipo, para comparar el rendimiento individual de cada club.
