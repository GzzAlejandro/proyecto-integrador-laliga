La columna Float con solo enteros es la que justifica por qué campos como overall_rating aparecen como Integer en el diccionario aunque pandas los lea como float64. Las que dicen “Sí” son esas.

```python
tablas = {
    "Match": df1,
    "Player_Attributes": df2,
    "Player": df3,
    "Team_Attributes": df4,
    "Team": df5,
}

filas = []
for nombre, df in tablas.items():
    for col in df.columns:
        s = df[col]
        solo_enteros = s.dtype.kind == "f" and (s.dropna() % 1 == 0).all()
        filas.append({
            "Tabla": nombre,
            "Campo": col,
            "Tipo pandas": str(s.dtype),
            "Float con solo enteros": "Sí" if solo_enteros else "No",
            "Nulos": s.isna().sum(),
            "% nulos": round(s.isna().mean() * 100, 1),
            "Únicos": s.nunique(),
            "Ejemplo": str(s.dropna().iloc[0])[:40] if s.notna().any() else "",
        })

respaldo = pd.DataFrame(filas)

# Ver una tabla a la vez
respaldo[respaldo["Tabla"] == "Match"]
```
