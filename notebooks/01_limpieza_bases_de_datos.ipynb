```python
# Importacion de Librerias

import pandas as pd
```

```python
# Usando la ruta esto permite que python busque el documento en otra carpeta sin exponer nuestras rutas privadas

ruta_match = '../data/raw/Match.csv'
df_partidos = pd.read_csv(ruta_match)
```

```python
df_partidos.head()
```

```python
# Al ser una base de datos relacional podemos filtrar los partidos a solo la liga espanola usando el country id o league id
df_spain = df_partidos[df_partidos['league_id'] == 21518].copy()
```

```python
print(df_spain)
```

# Ahora que tenemos la liga procedemos con el proceso de limpieza en las demas tablas

```python
# Filtrar equipos de la liga espanola por ID
equipos_ids = pd.concat([
    df_spain['home_team_api_id'], 
    df_spain['away_team_api_id']
]).unique()

df_equipo = pd.read_csv('../data/raw/Team.csv')
# Nos quedamos solo con los equipos que jugaron en España
df_equipos_laliga = df_equipo[df_equipo['team_api_id'].isin(equipos_ids)]
```

```python
# Revisamos la correcta ejecucion
print(f"Equipos españoles filtrados: {len(df_equipos_laliga)}")
```

```python
df_equipos_laliga.head()
```

```python
# Mismo proceso para los atributos de equipo
df_atributos_equipos = pd.read_csv('../data/raw/Team_Attributes.csv')
```

```python
# Filtramos solo quedandonos con los atributos de equipos de la liga

df_atributos_equipos_laliga = df_atributos_equipos[df_atributos_equipos['team_api_id'].isin(equipos_ids)]

```

```python
# Comprobación de limpieza
print(f"Registros totales en Europa: {len(df_atributos_equipos)}")
print(f"Registros tácticos filtrados para España: {len(df_atributos_equipos_laliga)}")
display(df_atributos_equipos_laliga.head())
```

# Por ultimo vamos a anadir a los jugadores, una limitacion de este dataset es que solo nos registra a los titulares, sin embargo el motivo por el que lo vamos a limtiar igualmente a estos sera por eficiencia de memoria, tener a toda la lista de jugadores volvera nuestro modelo mucho mas pesado de forma sumamente inecesaria, es mejor limitarlo un poco a tener que estar adivinando sin pruebas claras quien es suplente de cada equipo

```python
df_jugadores = pd.read_csv('../data/raw/Player.csv')
```

```python
df_jugadores.head()
```

```python
# Aplicaremos una condicion a los 22 jugadores titulares para crear una columna jugadores
columnas_jugadores = [f'home_player_{i}' for i in range(1, 12)] + [f'away_player_{i}' for i in range(1, 12)]

# Extraemos los ID
jugadores_ids = pd.unique(df_spain[columnas_jugadores].values.ravel()) #.Ravel vuelve todo una lista plana

# Llevamos a cabo una limpieza de nulos

jugadores_ids = jugadores_ids[~pd.isna(jugadores_ids)]

```

```python
# Cargamos la df de jugadores y atributos nuevamente (para mantener un orden)

df_jugadores = pd.read_csv('../data/raw/Player.csv')
df_atributos_jugadores = pd.read_csv('../data/raw/Player_Attributes.csv')
```

```python
# Filtramos ambas tablas usando nuestra lista de IDs de La Liga
df_jugadores_laliga = df_jugadores[df_jugadores['player_api_id'].isin(jugadores_ids)]
df_atributos_jugadores_laliga = df_atributos_jugadores[df_atributos_jugadores['player_api_id'].isin(jugadores_ids)]
```

```python
# Comprobamos que la limpieza fue correcta

# Comprobación del impacto de la limpieza
print(f"Total de jugadores únicos en La Liga: {len(df_jugadores_laliga)}")
print(f"Registros de atributos totales (Europa): {len(df_atributos_jugadores)}")
print(f"Registros de atributos filtrados (España): {len(df_atributos_jugadores_laliga)}")
```

```python
df_atributos_jugadores_laliga.head()
```

```python
# Exportacion de datos a nuestra carpeta de datos procesados para llevar a cabo su manipulacion futura
df_spain.to_csv('../data/processed/Match_LaLiga.csv', index=False)
df_equipos_laliga.to_csv('../data/processed/Team_LaLiga.csv', index=False)
df_atributos_equipos_laliga.to_csv('../data/processed/Team_Attributes_LaLiga.csv', index=False)
df_jugadores_laliga.to_csv('../data/processed/Player_LaLiga.csv', index=False)
df_atributos_jugadores_laliga.to_csv('../data/processed/Player_Attributes_LaLiga.csv', index=False)
```
