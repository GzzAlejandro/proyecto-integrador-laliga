```python
# Importacion de Librerias

import pandas as pd
```

```python
# Usando la ruta esto permite que python busque el documento en otra carpeta sin exponer nuestras rutas privadas

ruta_match = '../data/raw/Match.csv'
df_partidos = pd.read_csv(ruta_match)
```

```python
df_partidos.head()
```

```python
# Al ser una base de datos relacional podemos filtrar los partidos a solo la liga espanola usando el country id o league id
df_spain = df_partidos[df_partidos['league_id'] == 21518].copy()
```

```python
print(df_spain)
```

# Ahora que tenemos la liga procedemos con el proceso de limpieza en las demas tablas

```python
# Filtrar equipos de la liga espanola por ID
equipos_ids = pd.concat([
    df_partidos['home_team_api_id'], 
    df_partidos['away_team_api_id']
]).unique()

df_equipo = pd.read_csv('../data/raw/Team.csv')
# Nos quedamos solo con los equipos que jugaron en España
df_equipos_laliga = df_equipo[df_equipo['team_api_id'].isin(equipos_ids)]
```

```python
# Revisamos la correcta ejecucion
print(f"Equipos españoles filtrados: {len(df_equipos_laliga)}")
```
